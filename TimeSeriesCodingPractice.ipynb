```python
import altair as alt
import pandas as pd
from vega_datasets import data
```

## Load Seattle Weather Dataset

```python
# Load data
df = data.seattle_weather()

# Ensure date column is a datetime object
#TODO
df['date'] = pd.to_datetime(df['date'])
# Print the min and max dates in the dataset
#TODO
min_d = df['date'].min()
max_d = df['date'].max()
print('min date:', min_d, ' Max Date: ' , max_d )
```

## Visualize Temp Max Over Time

Create a line chart of `temp_max` over time. Include a title and a tooltip showing date and temp.

```python
#TODO
temp_max = alt.Chart(df).mark_line().encode(
    alt.X('date', axis=alt.Axis(format='%Y', tickCount='year')),
    alt.Y('temp_max'),
    tooltip=['temp_max', 'date']
).properties(width=700, title="Max temperature over time")
temp_max
```

## Create a Precipitation Bar Chart

1. Create a bar chart of precipitation over time.
2. Vertically concat your bar chart with your line chart from the previous exercise. The line graph should be on top, bar graph on bottom.
3. Make the x-axis shared.

```python
#TODO
precipitation = alt.Chart(df).mark_bar().encode(
    alt.X('date', axis=alt.Axis(format='%Y', tickCount='year')),
    alt.Y('precipitation'),
).properties(width=700)
combined = alt.vconcat(
    temp_max,
    precipitation
).resolve_scale(x='shared')

combined
```

## Adding Brush Interactivity

1. Add an interval selection (brush) to the bar chart from the previous exercise.
2. Modify the temp line graph so it responds to the brush.
3. Vertically concat the two graphs (line chart on top, bar chart on bottom).
4. Make the x-axis shared.

```python
#TODO
brush = alt.selection_interval(encodings=['x'])
(temp_max.transform_filter(brush) & precipitation.add_params(brush)).resolve_scale(x='shared')
```

## Visualize Temp Max Overlaid by 7 Day Rolling Average

1. Compute a 7-day rolling average of temp_max.
2. Overlay the rolling average line graph on top of the original temp line graph. Use a different color and strokeDash for the two lines so they are distinguishable.
3. Include a tooltip for the rolling average line.

```python

df = df.sort_values('date')
df['temp_max_7d'] = df['temp_max'].rolling(window=7, min_periods=1).mean()

temp_max = alt.Chart(df).mark_line().encode(
    alt.X('date', axis=alt.Axis(format='%Y', tickCount='year')),
    alt.Y('temp_max'),
)

line_roll = alt.Chart(df).mark_line(color='red').encode(
    alt.X('date'),
    alt.Y('temp_max_7d', title='7-day avg temp_max'),
    tooltip=['temp_max_7d', 'date']
)

(temp_max + line_roll).properties(width=700, title="Temp max with 7-day rolling average")
```

```python
brush = alt.selection_interval(encodings=['x'])

precipitation= precipitation.encode(
    color=alt.when(brush).then(alt.ColorValue('#4c78a8')).otherwise(alt.ColorValue('grey')))

(temp_max.transform_filter(brush) & precipitation.add_params(brush)).resolve_scale(x='shared')
```
